```python
import matplotlib.pyplot as plt
```

```python
# bar chart
categories = ['A', 'B', 'C', 'D', 'E']
values = [10, 24, 38, 18, 27]

plt.figure(figsize = (6, 4))
plt.bar(categories, values, color = "brown", edgecolor = "black")
plt.title("Bar chart", fontsize = 20)
plt.xlabel("Categories", fontsize = 16)
plt.ylabel("Values", fontsize = 16)
plt.tight_layout()
plt.show()
```

```python
# Line Chart
months = ['Jan', 'Feb', 'March', 'Apr', 'May', 'June']
sales = [150, 200, 190, 220, 300, 250]

plt.figure(figsize=(10, 4))
plt.plot(months, sales, marker = "o", linewidth = 3, markersize = 10, markerfacecolor = "black", markeredgecolor = "black")
plt.title("line Chart", fontsize = 20)
plt.xlabel("Months", fontsize = 16)
plt.ylabel("Sales", fontsize = 16)
plt.tight_layout()
plt.grid(True)
plt.show()
```

```python
# Area Chart
months = ['Jan', 'Feb', 'March', 'Apr', 'May', 'June']
sales = [150, 200, 190, 220, 300, 250]

plt.figure(figsize=(10, 4))
plt.fill_between(months, sales, color = "skyblue", alpha = 0.5)
plt.plot(months, sales, marker = "o", linewidth = 3, markersize = 10, markerfacecolor = "black", markeredgecolor = "black")
plt.title("Area Chart", fontsize = 20)
plt.xlabel("Months", fontsize = 16)
plt.ylabel("Sales", fontsize = 16)
plt.tight_layout()
plt.grid(False)
plt.show()
```

```python
# Pie chart
categories = ['Apple','Bananas', 'Cherries', 'Grapes']
values = [25, 35, 20, 20]

plt.figure(figsize = (10, 4))
plt.pie(values, labels = categories, autopct = '%1.1f%%')
plt.title("Pie chart distribution", fontsize = 20)
plt.show()
```

```python
# Donut chart
categories = ['Apple','Bananas', 'Cherries', 'Grapes']
values = [25, 35, 20, 20]

plt.figure(figsize = (10, 4))
plt.pie(values, labels = categories, autopct = '%1.1f%%')
centre_circle = plt.Circle((0, 0), 0.40, fc = "white")
fig = plt.gcf()
fig.gca().add_artist(centre_circle)
plt.title("Pie chart distribution", fontsize = 20)
plt.show()
```

```python
# Histogram
import numpy as np
data = np.random.normal(50, 10, 200)
plt.figure(figsize = (10, 4))
plt.hist(data, bins =10, color = "orange", edgecolor = "black")
plt.title("Histogram", fontsize = 20)
plt.xlabel("Values", fontsize = 16)
plt.ylabel("Frequencies", fontsize = 16)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
```

```python
data = sns.load_dataset('tips')
data.head()
```

```python
data.shape
```

```python
data.describe
```

```python
# Box plot
plt.figure(figsize = (10, 4))
sns.boxplot(x = 'day', y = 'total_bill', data = data, palette = 'pastel')
plt.title('Box plot disttribution', fontsize = 20)
plt.show()
# these outside values -> outliers
# box middle line -> median(q2) - 50%
# box upper line -> max value - quantile(q3) - 75%
# box lower line -> min value - quantile(q1) - 25%
```

```python
# median value for total_bill of Thru
data[data['day'] == 'Thur']['total_bill'].median()
```

```python
q2 = data.loc[data['day'] == 'Thur', 'total_bill'].quantile(0.50)
q2
```

```python
q1 = data.loc[data['day'] == 'Thur', 'total_bill'].quantile(0.25)
q1
```

```python
q3 = data.loc[data['day'] == 'Thur', 'total_bill'].quantile(0.75)
q3
```

```python
# Violin Plot - show destribution along the median
plt.figure(figsize = (10, 4))
sns.violinplot(x = 'day', y = 'total_bill', data = data, palette = 'Set1')
plt.title('Violin plot disttribution', fontsize = 20)
plt.show()
```

```python
# on thur mostly people pay that same amount (median)
```

```python
# Density Plot - smooth curev (KDE) shows the data distribution for a single numerical column
plt.figure(figsize = (10, 4))
sns.kdeplot(data['total_bill'], fill = True, color = "orange")
plt.title('Density plot disttribution', fontsize = 20)
plt.show()
```

```python
plt.figure(figsize = (10, 4))
sns.kdeplot(data['total_bill'],label = "Total Bill", fill = True, color = "orange")
sns.kdeplot(data['tip'], label = "Tip", fill = True, color = "pink")
plt.title("KDE plot curev", fontsize = 20)
plt.xlabel('Total Bills', fontsize = 16)
plt.ylabel('Tip', fontsize = 16)
plt.legend()
plt.show()
```

```python
# Scatter plot - shows corelation or relationship between 2 coloumns (numeric values)
plt.figure(figsize = (10, 4))
plt.scatter(data['total_bill'], data['tip'], color = "brown", s = 100, edgecolor = "black")
plt.title("Scatter Plot", fontsize = 20)
plt.xlabel('Total Bills', fontsize = 16)
plt.ylabel('Tip', fontsize = 16)
plt.tight_layout()
plt.show()
```
